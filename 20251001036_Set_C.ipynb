```python
#Q1- take input of kc and write and message for invalid input
kc=input("Enter the value of kc:")
print("The value of Kc is:",kc)
print("The data type kc is:",type(kc))

#convert kc to float and report if the kc<0
try:
    kc_new = float(kc)

    if kc_new <= 0:
        print("Invalid input. Please enter a positive value for kc.")
    else:
        print("The data type of converted Kc is:", type(kc_new))

        # Coefficients for: kc*x^2 - (kc + 1)*x + 0.25*kc = 0
        a = kc_new
        b = -(kc_new + 1)
        c = 0.25 * kc_new

        print("The value of a is:", a)
        print("The value of b is:", b)
        print("The value of c is:", c)

        discriminant = b**2 - 4 * a * c
        print("The value of discriminant is:", discriminant)

        if discriminant < 0:
            print("The roots are complex.")
        else:
            root1 = (-b + discriminant**0.5) / (2 * a)
            root2 = (-b - discriminant**0.5) / (2 * a)
            print("The roots are real:", root1, "and", root2)

            valid_roots = [root for root in (root1, root2) if 0 <= root <= 0.5]

            if not valid_roots:
                print("No valid roots in the range 0 to 0.5.")
            else:
                for root in valid_roots:
                    A = 0.5 - root
                    B = 0.5 - root
                    C = root
                    conversion_percent = (root / 0.5) * 100

                    print(f"Valid root: {root}")
                    print(
                        f"Equilibrium concentrations: "
                        f"A = {A:.3f}, B = {B:.3f}, C = {C:.3f}"
                    )
                    print(f"Conversion: {conversion_percent:.2f}%")

                    if conversion_percent >= 70:
                        print("The conversion percent is High.")
                    elif conversion_percent >= 40:
                        print("The conversion percent is Moderate.")
                    else:
                        print("The conversion percent is Low.")

except ValueError:
    print("Invalid input. Please enter a numeric value for kc.")
```

```python
#Use Eulers methos C_new = C_old - (k*c_old*delta_t) for dc/dt = -k*c
#Intial value of C_o = 1 , k=0.12, delta_t=0.5
#Using a loop simulate 5 minutes and print a table for Time(min), Concentration(mol/L) and conversion(%)
C_old = 1.0
k = 0.12
delta_t = 0.5
print("\nTime(min)\tConcentration(mol/L)\tConversion(%)")
for time in range(0, 6):  # Simulate for 5 minutes (0, 1, 2, 3, 4, 5)
    C_new = C_old - (k * C_old * delta_t)
    conversion = (1 - C_new) * 100
    print(f"{time}\t\t{C_new:.4f}\t\t\t{conversion:.2f}")
    C_old = C_new 

#Using math.exp() calculate the exact value of C_new and compare it with the value obtained from Eulers method and print the absolute error of euler method
import math
math_C_new = 1 * math.exp(-k * 5)  # Exact value after 5 minutes
absolute_error = abs(math_C_new - C_new)
print(f"\nExact value of C_new after 5 minutes using math.exp(): {math_C_new:.4f}")
print(f"Absolute error of Euler's method: {absolute_error:.4f}")

#Using while loop repeat the simulation till conversion reaches 70% or time reaches 100%. Print time needed and the final conversion
Conversion_percent = 0
C_old = 1.0
time = 0
while Conversion_percent < 70 or time < 100:
    C_new = C_old - (k * C_old * delta_t)
    Conversion_percent = (1 - C_new) * 100
    time += delta_t
    C_old = C_new
print(f"\nTime needed for 70% conversion: {time:.2f} minutes")
print(f"Final conversion: {Conversion_percent:.2f}%")
```

```python
#QUESTION 3 FEED BLEND SCREENING AND SAMPLE SCHEDULE:
#PART A
#a plant has n candidate feed component and wants to blend r of them(in combination so order does not matter)
#Write a code to define a function which calculates factorial in recursive format
def factorial(x):
    if x == 0 or x == 1:
        return 1
    else:
        return x*factorial(x-1)
```

```python
#QUESTION 3 FIBONACCI SEQUENCE
#PART B
#During a cooling experiment, temperature readings ar taken at a time = Fibonacci number * 3 min.
#Create a itterative function fib_iter(n) that returens a list of n fibonacci number in a list
def fib_iter(n):
    numbers = []
    a, b = 0, 1

    for _ in range(n):
        numbers.append(a)
        a, b = b, a + b

    return numbers
```

```python
#Write a code which n as input and calculates iterative function and returns the value of first n fibonacci in a list and calculate using function
#The table should be of format Reading no. | Fibonacci tag | Time (min)
n = int(input("Enter the number of n to generate: "))
if n <= 0:
    print("Please enter a positive integer for n.")
else:
    fibonacci_numbers = fib_iter(n)
    print("\nReading no.\tFibonacci tag\tTime (min)")
    for i, fib in enumerate(fibonacci_numbers):
        time = fib * 3
        print(f"{i + 1}\t\t{fib}\t\t{time}")
```

```python
#write fibonacci in recursive format and returs the fibanacci number at position k
# define function as fib_rec(k) not as fib_recursive(k) to avoid confusion with the iterative function
def fib_rec(k):
    if k <= 0:
        return 0
    elif k == 1:
        return 1
    else:
        return fib_rec(k - 1) + fib_rec(k - 2)
```
