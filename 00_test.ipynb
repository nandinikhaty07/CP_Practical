```python
#get marks of a student and give grade based on marks
#95 and above - grade A , 85 and above - grade B, 75 and above - grade C, otherwise - grade D
marks = int(input("Enter your marks:"))
if marks >= 95:
    print("Grade A")
elif marks >= 85:
    print("Grade B")
elif marks >= 75:
    print("Grade C")
else:
    print("Grade D")
    
```
